## Acess AzureML SDK
Through the python SDK, you can interact with all aspects of the AzureML workspace, as you can through the GUI or CLI

```python
from azure.ai.ml import MLClient
from azure.identity import DefaultAzureCredential
```

```python
ml_client = MLClient.from_config(credential=DefaultAzureCredential())
```

```python
ml_client
```

In this example we demonstrate interacting with the compute resources, list all the compute instances or clusters and their type.

```python
pandas.DataFrame.from_records([ {'name':c1.name, 'type': c1.type} for c1 in ml_client.compute.list()])
```
