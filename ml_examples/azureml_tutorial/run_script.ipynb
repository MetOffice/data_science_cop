# Run a script on AzureML Compute

In this tutoriall, we demonstrate running a basic script on different sorts of compute.

```python
import pathlib
```

```python
from azureml.core import Workspace, Experiment, Environment, ScriptRunConfig

```

```python
3+2
```

```python
ws = Workspace.from_config()
```

```python
ws
```

```python
experiment = Experiment(workspace=ws, name='day1-experiment-hello')
experiment
```

```python
# Select this for a compute instance
compute_target = 'dscoptest1'

# Select this for a CPU based c ompute cluster
# compute_target = 'dscopgpucluster'

#Select this for a GPU based compute cluster
# compute_target = 'dscopgpucluster'
```

```python
config = ScriptRunConfig(source_directory='.', script='hello.py', compute_target=compute_target)
config
```

```python
?config
```

```python
run = experiment.submit(config)
run
```

```python
aml_url = run.get_portal_url()
aml_url
```
